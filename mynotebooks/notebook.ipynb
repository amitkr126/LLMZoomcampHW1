```python
# Run just the search portion
search_results = assistant.search(query)

# Print the filename of the very first result
print("The first result filename is:", search_results[0]['filename'])
```

```python
# Unpack the tuple returned by .rag()
answer, usage = assistant.rag(query)

print("--- ANSWER ---")
print(answer)
print("\n--- TOKEN USAGE ---")
print(f"Input Tokens (Prompt): {usage.prompt_tokens}")
print(f"Output Tokens (Completion): {usage.completion_tokens}")
print(f"Total Tokens: {usage.total_tokens}")
```
